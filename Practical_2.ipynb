```python
# Program 1: Right-Angled Triangle
for i in range(1, 6):
  for j in range(1, i + 1):
    print("*", end=" ")
  print()
```

```python
# Program 2: Inverted Triangle
for i in range(5, 0, -1):
  for j in range(1, i + 1):
    print("*", end=" ")
  print()
```

```python
# Program 3: Number Pattern
for i in range(1, 6):
  for j in range(1, i + 1):
    print(j, end=" ")
  print()
```

```python
# Program 4: Pyramid Pattern
rows = 5
for i in range(1, rows + 1):
  print(" " * (rows - i), end="")  # Spaces
  print("* " * i)
```

```python
# Program 6: Alphabet Triangle
ch = 65  # ASCII for 'A'
for i in range(5):
  for j in range(i + 1):
    print(chr(ch), end=" ")
  ch += 1
  print()
```

```python
# Program 7: Floyd's Triangle
num = 1
for i in range(1, 6):
  for j in range(1, i + 1):
    print(num, end=" ")
  num += 1
  print()
```

```python
# Program 8: Diamond Pattern
rows = 5
# Upper part
for i in range(1, rows + 1):
  print(" " * (rows - i) + "* " * i)
  # Lower part
for i in range(rows - 1, 0, -1):
    print(" " * (rows - i) + "* " * i)
```
